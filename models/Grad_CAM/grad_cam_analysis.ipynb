# Grad-CAM analysis

This notebook runs the simple Grad-CAM workflow for the available repository checkpoints. Run it from the repository root.

## 1. Editable settings

- Use `smoke` first for one model and one image.
- Change to `full` for all 15 shared images.
- CPU is slow but reliable.

```python
RUN_MODE = "smoke"  # Change to "full" after the smoke test passes
SMOKE_MODEL = "resnet50_fgvc_pmg_isqrtcov"
DEVICE = "cpu"      # Other choices: "mps", "cuda", "auto"
RESULTS_DIR = "results/grad_cam"
```

## 2. Import the local Grad-CAM code

```python
import sys
from pathlib import Path

grad_cam_folder = Path("models/Grad_CAM").resolve()
if str(grad_cam_folder) not in sys.path:
    sys.path.insert(0, str(grad_cam_folder))

from grad_cam_utils import get_model_inventory, select_shared_samples
from run_grad_cam import run_workflow
```

## 3. Check model availability

Missing checkpoints are listed instead of being replaced with other weights.

```python
inventory = get_model_inventory()
inventory[["model_key", "checkpoint_exists", "inventory_status", "target_layer"]]
```

## 4. Check the 15 shared samples

```python
selected_samples = select_shared_samples(RESULTS_DIR)
selected_samples
```

## 5. Run Grad-CAM

The script first verifies model logits and then creates predicted-class and, when needed, true-class Grad-CAM figures.

```python
summary = run_workflow(
    mode=RUN_MODE,
    device_name=DEVICE,
    results_dir=RESULTS_DIR,
    smoke_model=SMOKE_MODEL,
)
summary
```
